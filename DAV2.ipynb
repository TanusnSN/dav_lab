```python
import pandas as pd
import numpy as np
#Create/load the dataset
data= {
    "Name": ["Amit","Bebo","Chethan","Amit","Divya","Ravi","Sneha","Karan",
             "Pooja","Manish","Anjali","Suresh","Kavya","Rahul","Meera","Vikram","Nisha","Arjun","Divya","Rohit","Swathi","Naveen","Priya","Karan","Deepa"]
    , "Age": [25,np.nan,30,25,28,22,24,np.nan,27,26,23,29, np.nan,25,22,28,24,np.nan,28,26,23,30,27,np.nan,25]
    , "Marks": [85,90,np.nan,85,78,95,99,72,np.nan,91,67,84,79,np.nan,93,88,76,82,78,np.nan,90,85,71,72,89]
    , "City" : ["Mysore", "Mysore", "Bangalore", "Mysore",np.nan,"Delhi","Mysore",
                "Mangalore","Bangalore",np.nan,"Mysore","Hubli","Mangalore",
                "Delhi","Bangalore",np.nan,"Mysore","Hubli",np.nan,"Delhi"
                ,"Mangalore","Mysore","Bengalore","Mangalore","Hubli"]
     }

df=pd.DataFrame(data)
print("Dataset:")
print(df)
#Basic Information
print("\nShape of dataset (rows, columns):",df.shape)
print("\nColumn names:",df.columns.tolist())
print("\nData types of each columns:")
print(df.dtypes)
print("\nDataset Info:")
df.info()
#Descriptive Statistics

print("\nDescriptive Statistics (numerical columns):")
print(df.describe())
print("\nMean:\n", df.mean(numeric_only=True))
print("\nMedian:\n", df.median(numeric_only=True))
print("\nStandard Deviation:\n", df.std(numeric_only=True))
#Identifying Missing Values

print("\nMissing Values in each column:")
print(df.isnull().sum())
print("\nTotal missing values in dataset:", df.isnull().sum().sum())
#Identifying Duplicate Records

print("\nDuplicate rows in dataset:")
print(df[df.duplicated()])
print("\nNumber of duplicate rows:", df.duplicated().sum())
print("\nEDA process completed successfully!")
```
